```python

import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
def wireframe_trace(X, Y, Z, color='blue', name='', step=10):
    """Build wireframe lines (along x and y directions) for a surface."""
    lines = []
    # lines along y-direction (constant x rows)
    for i in range(0, X.shape[0], step):
        lines.append(go.Scatter3d(
            x=X[i, :], y=Y[i, :], z=Z[i, :],
            mode='lines', line=dict(color=color, width=2),
            name=name, showlegend=False
        ))
    # lines along x-direction (constant y columns)
    for j in range(0, X.shape[1], step):
        lines.append(go.Scatter3d(
            x=X[:, j], y=Y[:, j], z=Z[:, j],
            mode='lines', line=dict(color=color, width=2),
            name=name, showlegend=False
        ))
    return lines

x = np.linspace(1,5, 100) #v105
y = np.linspace(2,6, 100) #v112 z136 exp()

X, Y = np.meshgrid(x, y)

# McCormick relaxation of z = x * (1/y)
# z>= 0.16666666666666666 * x +  1 * 1/y +  -0.16666666666666666
# z>= 0.5 * x +  5 * 1/y +  -2.5
# z<= 0.16666666666666666 * x +  5 * 1/y +  -0.8333333333333333
# z<= 0.5 * x +  1 * 1/y +  -0.5
# McCormick relaxation of x = z * y)
# z>= 0.5 * x +  -1.25 * y +  2.5
# z>= 0.16666666666666666 * x +  -0.027777777777777776 * y +  0.16666666666666666
# z<= 0.5 * x +  -0.08333333333333333 * y +  0.16666666666666666
# z<= 0.16666666666666666 * x +  -0.4166666666666667 * y +  2.5
Z = X / Y
z1=0.5*(X-2.5*Y+5)
z2=1/6*(X-1/6*Y+1)

z3=0.5*X+5*(1/Y)-2.5
z4=1/6*X+(1/Y)-1/6

fig = go.Figure()
# for trace in wireframe_trace(X, Y, Z, color='red', name='z', step=5):
#     fig.add_trace(trace)
# for trace in wireframe_trace(X, Y, z1, color='black', name='z1', step=5):
#     fig.add_trace(trace)
for trace in wireframe_trace(X, Y, z2, color='black', name='z2', step=5):
    fig.add_trace(trace)
# for trace in wireframe_trace(X, Y, z3, color='orange', name='z3', step=5):
#     fig.add_trace(trace)
for trace in wireframe_trace(X, Y, z4, color='orange', name='z4', step=5):
    fig.add_trace(trace)
fig.update_layout(
    title='Multiple wireframe surfaces',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'),
    width=800, height=600
)
fig.show()
```

```python
 
def mccormick_x_div_y(x_bounds, y_bounds):

    x_low, x_high = x_bounds
    y_low, y_high = y_bounds

    if y_low <= 0 <= y_high:
        raise ValueError("y interval must not contain 0 (1/y is undefined/discontinuous there).")
    if x_low > x_high or y_low > y_high:
        raise ValueError("Each bound pair must be given as (low, high) with low <= high.")


    inv_y_low=1/y_high
    inv_y_high=1/y_low

    print("McCormick relaxation of z = x * (1/y)")
    print("z>=",inv_y_low,"* x + ",x_low, "* 1/y + ",-x_low * inv_y_low)
    print("z>=",inv_y_high,"* x + ",x_high, "* 1/y + ",-x_high * inv_y_high)
    print("z<=",inv_y_low,"* x + ",x_high, "* 1/y + ",-x_high * inv_y_low)
    print("z<=",inv_y_high,"* x + ",x_low, "* 1/y + ",-x_low * inv_y_high)

def mccormick_z_mul_y(x_bounds, y_bounds):

    x_low, x_high = x_bounds
    y_low, y_high = y_bounds

    z_low=min(x_low/y_low,x_low/y_high,x_high/y_low,x_high/y_high)
    z_high=max(x_low/y_low,x_low/y_high,x_high/y_low,x_high/y_high)

    if y_low <= 0 <= y_high:
        raise ValueError("y interval must not contain 0 (1/y is undefined/discontinuous there).")
    if x_low > x_high or y_low > y_high:
        raise ValueError("Each bound pair must be given as (low, high) with low <= high.")

    print("McCormick relaxation of x = z * y)")
    if y_low>0:
        print("z>=",1/y_low, "* x + ", -z_high/y_low,"* y + ",z_high)
        print("z>=",1/y_high, "* x + ", -z_low/y_high,"* y + ",z_low)
        print("z<=",1/y_low, "* x + ", -z_low/y_low,"* y + ",z_low)
        print("z<=",1/y_high, "* x + ", -z_high/y_high,"* y + ",z_high)
    else:
        print("z<=",1/y_low, "* x + ", -z_high/y_low,"* y + ",z_high)
        print("z<=",1/y_high, "* x + ", -z_low/y_high,"* y + ",z_low)
        print("z>=",1/y_low, "* x + ", -z_low/y_low,"* y + ",z_low)
        print("z>=",1/y_high, "* x + ", -z_high/y_high,"* y + ",z_high)
```
