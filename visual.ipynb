```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
def wireframe_trace(X, Y, Z, color='blue', name='', step=10):
    """Build wireframe lines (along x and y directions) for a surface."""
    lines = []
    # lines along y-direction (constant x rows)
    for i in range(0, X.shape[0], step):
        lines.append(go.Scatter3d(
            x=X[i, :], y=Y[i, :], z=Z[i, :],
            mode='lines', line=dict(color=color, width=2),
            name=name, showlegend=False
        ))
    # lines along x-direction (constant y columns)
    for j in range(0, X.shape[1], step):
        lines.append(go.Scatter3d(
            x=X[:, j], y=Y[:, j], z=Z[:, j],
            mode='lines', line=dict(color=color, width=2),
            name=name, showlegend=False
        ))
    return lines
def x_div_y(x_bounds, y_bounds):

    x_low, x_high = x_bounds
    y_low, y_high = y_bounds

    if y_low <= 0 <= y_high:
        raise ValueError("y interval must not contain 0 (1/y is undefined/discontinuous there).")
    if x_low > x_high or y_low > y_high:
        raise ValueError("Each bound pair must be given as (low, high) with low <= high.")


    inv_y_low=1/y_high
    inv_y_high=1/y_low

    print("Relaxation of z = x * (1/y)")
    print("z>=",inv_y_low,"* x + ",x_low, "* 1/y + ",-x_low * inv_y_low)
    print("z>=",inv_y_high,"* x + ",x_high, "* 1/y + ",-x_high * inv_y_high)
    print("z<=",inv_y_low,"* x + ",x_high, "* 1/y + ",-x_high * inv_y_low)
    print("z<=",inv_y_high,"* x + ",x_low, "* 1/y + ",-x_low * inv_y_high)

def z_mul_y(x_bounds, y_bounds):

    x_low, x_high = x_bounds
    y_low, y_high = y_bounds

    z_low=min(x_low/y_low,x_low/y_high,x_high/y_low,x_high/y_high)
    z_high=max(x_low/y_low,x_low/y_high,x_high/y_low,x_high/y_high)

    if y_low <= 0 <= y_high:
        raise ValueError("y interval must not contain 0 (1/y is undefined/discontinuous there).")
    if x_low > x_high or y_low > y_high:
        raise ValueError("Each bound pair must be given as (low, high) with low <= high.")

    print("Relaxation of x = z * y")
    if y_low>0:
        print("z>=",1/y_low, "* x + ", -z_high/y_low,"* y + ",z_high)
        print("z>=",1/y_high, "* x + ", -z_low/y_high,"* y + ",z_low)
        print("z<=",1/y_low, "* x + ", -z_low/y_low,"* y + ",z_low)
        print("z<=",1/y_high, "* x + ", -z_high/y_high,"* y + ",z_high)
    else:
        print("z>=",1/y_low, "* x + ", -z_low/y_low,"* y + ",z_low)
        print("z>=",1/y_high, "* x + ", -z_high/y_high,"* y + ",z_high)
        print("z<=",1/y_low, "* x + ", -z_high/y_low,"* y + ",z_high)
        print("z<=",1/y_high, "* x + ", -z_low/y_high,"* y + ",z_low)

def plot_wireframe(X, Y, Z,z1,z2,z3,z4,z_flag=True,z1_flag=True,z2_flag=True,z3_flag=True,z4_flag=True):
    fig = go.Figure()
    if z_flag:
        for trace in wireframe_trace(X, Y, Z, color='black', name='z', step=5): # black: original surface
            fig.add_trace(trace)
    if z1_flag:
        for trace in wireframe_trace(X, Y, z1, color='blue', name='z1', step=5): # blue: x/y relaxation
            fig.add_trace(trace)
    if z2_flag:
        for trace in wireframe_trace(X, Y, z2, color='blue', name='z2', step=5): # blue: x/y relaxation
            fig.add_trace(trace)
    if z3_flag:
        for trace in wireframe_trace(X, Y, z3, color='red', name='z3', step=5): # red: 1/y relaxation
            fig.add_trace(trace)
    if z4_flag:
        for trace in wireframe_trace(X, Y, z4, color='red', name='z4', step=5): # red: 1/y relaxation
            fig.add_trace(trace)
    fig.update_layout(
        title='Multiple wireframe surfaces',
        scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'),
        width=800, height=600
    )
    fig.show()
```

```python
z_mul_y((1, 5), (-6, -2))
x_div_y((1, 5), (-6, -2))

x = np.linspace(1,5, 100) #v105
y = np.linspace(-6,-2, 100) #v112 z136 exp()

X, Y = np.meshgrid(x, y)


Z = X / Y
z1= -0.16666666666666666 * X +  -0.4166666666666667 * Y +  -2.5
z2= -0.5 * X +  -0.08333333333333333 * Y +  -0.16666666666666666


z3= -0.16666666666666666 * X +  5 * 1/Y +  0.8333333333333333
z4= -0.5 * X +  1 * 1/Y +  0.5
plot_wireframe(X, Y, Z,z1,z2,z3,z4,True,True,False,True,False)
```
